```python
# Fill missing values
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Optional: Drop last_review (not needed now)
df = df.drop(columns=['last_review'])
```

```python
df=df[df['price']>0]
df=df[df['price']<1000]
```

```python
import numpy as np
df['price_per_preview']= np.where(
    df['price']==0,
    0,
    df['price']/df['number_of_reviews']

)
```

```python
df['name']=df['name'].fillna("unknown")
df['host_name']=df['host_name'].fillna("unknown")
```

```python
import sqlite3
conn=sqlite3.connect("airbnb.db")
```

```python
df.to_sql("listings",conn,if_exists='replace',index='False')
```

```python
df.info()
```

```python
# Expensive areas

query = """
SELECT neighbourhood_group,
       AVG(price) as avg_price
FROM listings
GROUP BY neighbourhood_group
ORDER BY avg_price DESC
LIMIT 5

"""
df_expensive_neighbourhood = pd.read_sql(query,conn)
print(df_expensive_neighbourhood)
```

```python
# Which room types earn more

query2= """
SELECT room_type,
        AVG(price) as avg_price
FROM listings
GROUP BY room_type
ORDER BY avg_price DESC

"""
df_expensive_room_type=pd.read_sql(query2,conn)
print(df_expensive_room_type)

```

```python
# Do expensive listings get fewer reviews?
query3= """
SELECT price,
       number_of_reviews
FROM listings

"""
df_priceVSreview = pd.read_sql(query3,conn)
print(df_priceVSreview.head(3))


```
