````python
#import all necessary labraries
import requests
from bs4 import BeautifulSoup
import re
import pandas as pd
import numpy as np
````

## **Scrapping Movies Data**

````python
headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
    }
def fetch_movie_urls(url):

    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        return "Failed to retrieve the page, status code:", response.status_code
    soup = BeautifulSoup(response.text, 'html.parser')
    return soup


url = 'https://www.justwatch.com/in/movies?release_year_from=2000'
soup=fetch_movie_urls(url)
print(soup.prettify())
````

## **Fetching Movie URL's**

````python
# Write Your Code here
# list of movie urls
url_list = []
for x in soup.find_all('a',class_="title-list-grid__item--link")[:51]:
  url_list.append('https://www.justwatch.com'+ x['href'])
print(url_list)
````

## **Scrapping Movie Title**

````python
# Write Your Code here
# movie titles

movie_title_uncleaned=[]

for link in url_list:

  try:
    response = requests.get(link,headers=headers)
    soup = BeautifulSoup(response.text,'html.parser')
    title=soup.find_all('h1')[0].text
  except Exception as e:
    title = 'NA'
    print(f"Error fetching title from {link}: {e}")
  movie_title_uncleaned.append(title)

import re

movie_title = [re.sub(r'\s*\(\d{4}\)', '', title).strip() for title in movie_title_uncleaned]
print(movie_title)
````

## **Scrapping release Year**

````python
# Write Your Code here
# Release year
release_year=[]
for link in url_list:
  try:
    response=requests.get(link,headers=headers)
    soup=BeautifulSoup(response.text,'html.parser')
    year=soup.find_all('span',class_="release-year")[0].text.strip("()")
  except Exception as e:
    year='NA'
    print(f"Error fetching title from {link}:{e}")
  release_year.append(year)
print(release_year)
````

## **Scrapping Genres**

````python
# Write Your Code here
#Scrapping Genres

genres=[]
for link in url_list:
  try:
    response=requests.get(link,headers=headers)
    soup=BeautifulSoup(response.text,'html.parser')
    for x in soup.find_all("div",class_="poster-detail-infos"):
      if x.find_all("h3")[0].text=="Genres":
        genre= x.find_all("span")[0].text
  except Exception as e:
    year='NA'
    print(f"Error fetching title from {link}:{e}")
  genres.append(genre)

print(genres)
````

## **Scrapping IMBD Rating**

````python
# Write Your Code here
#Rating

movie_rating=[]
for link in url_list:
  try:
    response = requests.get(link,headers=headers)
    soup=BeautifulSoup(response.text,'html.parser')
    rating=soup.find_all("div",class_="nowrap")[0].text.strip()
    rating = re.sub(r"\s*\(.*?\)", "", rating)
  except Exception as e:
    rating="NA"
    print(f"error fetching rating {url}:{e}")
  movie_rating.append(rating)
print(movie_rating)
````

## **Scrapping Runtime/Duration**

````python
# Write Your Code here
# runtime
movie_runtime = []

for link in url_list:
    try:
        response = requests.get(link, headers=headers)
        soup = BeautifulSoup(response.text, 'html.parser')
        time = "NA"

        for block in soup.find_all("div", class_="poster-detail-infos"):
            title_tag = block.find("h3")
            value_tag = block.find("div", class_="poster-detail-infos__value")

            if title_tag and value_tag and title_tag.text.strip() == "Runtime":
                time = value_tag.text.strip()
                break
          # Found runtime, stop looking

    except Exception as e:
        time = "NA"
        print(f"Error fetching time from {link}: {e}")

    movie_runtime.append(time)

print(movie_runtime)
````

## **Scrapping Age Rating**

````python
# Write Your Code here
# Age rating
age_rating=[]
for link in url_list:
  try:
    response=requests.get(link,headers=headers)
    soup=BeautifulSoup(response.text,'html.parser')
    cert= "NA"
    for block in soup.find_all("div",class_="poster-detail-infos"):
      title_tag = block.find("h3")
      value_tag = block.find("div",class_="poster-detail-infos__value")
      if title_tag and value_tag and title_tag.text.strip()=="Age rating":
        cert= value_tag.text.strip()
        break


  except Exception as e:
    cert="NA"
    print(f"Error fetching cert from {link}: {e}")
  age_rating.append(cert)
print(age_rating)
````

## **Fetching Production Countries Details**

````python
# Write Your Code here
# Country
production_country=[]
for link in url_list:
  try:
    response=requests.get(link,headers=headers)
    soup=BeautifulSoup(response.text,'html.parser')
    country= "NA"
    for block in soup.find_all("div",class_="poster-detail-infos"):
      title_tag = block.find("h3")

      value_tag = block.find("div",class_="poster-detail-infos__value")
      if title_tag and value_tag and title_tag.text.strip()=="Production country":
        country= value_tag.text.strip()
        break


  except Exception as e:
    country="NA"
    print(f"Error fetching country from {link}: {e}")
  production_country.append(country)
print(production_country)


````

## **Fetching Streaming Service Details**

````python
# Write Your Code here
# streaming_service

movie_stream_service_uncleaned=[]
for url in url_list:
  try:
     content=requests.get(url,headers=headers)
     soup=BeautifulSoup(content.text,'html.parser')
     names=[x['alt'] for x in soup.find_all('img',class_='provider-icon wide icon')]
  except:
    names='NA'
  movie_stream_service_uncleaned.append(" , ".join(names))

# Removing two just watches at the end

movie_stream_service = []

for entry in movie_stream_service_uncleaned:
    parts = [p.strip() for p in entry.split(',')]
    # Remove last two "JustWatch" if they are there
    count = 0
    while parts and parts[-1] == "JustWatch" and count < 2:
        parts.pop()
        count += 1
    movie_stream_service.append(', '.join(parts))
print(movie_stream_service)
````

## **Now Creating Movies DataFrame**

````python
# Write Your Code here
#
movie_dict={"movie_title":movie_title,"release_year":release_year,"genres":genres,"movie_rating":movie_rating,
            "movie_runtime":movie_runtime,"age_rating":age_rating,"production_country":production_country,
            "movie_stream_service":movie_stream_service}

movie_df = pd.DataFrame(movie_dict)
````
